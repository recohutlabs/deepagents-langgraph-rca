# Counterfactual Validation

This notebook compares fixture worlds through deterministic source tools. Live model comparison is optional.

```python
from deepagents_langgraph_rca.config import load_config
from deepagents_langgraph_rca.sources import DataSources

config = load_config(require_model=False)
fixtures = {
    'reference': None,
    'removed-access': 'counterfactual/removed-access',
    'reversed-access': 'counterfactual/reversed-access',
    'renamed-case': 'blind/renamed-case',
}

summary = {}
for name, fixture in fixtures.items():
    sources = DataSources(config, fixture=fixture)
    tasks = sources.list_tasks(due_before='2026-07-17')
    summary[name] = [(task['task_id'], task['status']) for task in tasks]
summary
```

The validation principle is evidence movement, not exact phrasing. Removed or reversed evidence should weaken or change the model's conclusion.
